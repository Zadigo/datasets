```python
import pandas
```

```python
df1 = pandas.read_json('countries_by_continent.json')
```

```python
df2 = pandas.read_json('fifa_codes.json')
```

```python
df3 = pandas.merge(df1, df2, on='country', how='inner')
```

```python
df3 = df3.rename(columns={'code': 'fifa'})
```

```python
df3.sort_values('country', inplace=True)
```

```python
df3.to_json('countries_by_continent_and_fifa.json', orient='records', indent=4)
```
