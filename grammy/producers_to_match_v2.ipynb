```python
import csv
import json
import pathlib
import pandas
```

```python
grammies = list(csv.reader(pathlib.Path('grammy_winners_v2.csv').open()))
```

```python
grammies[:2]
```

```python
with open(pathlib.Path('tmp/producers_to_match.json'), 'r') as f:
    producers_to_match = json.load(f)
```

```python
YEAR = 1970
```

```python
for item in producers_to_match:
    if item['producers'] is not None:
        for csv_value in grammies:
            if csv_value[-1] == 'False':
                continue

            if csv_value[1] == f'{YEAR}':
                logic = [
                    csv_value[3].lower() == item['award'].lower(),
                    (
                        csv_value[4].lower() == item['title'].lower() or
                        csv_value[5].lower() == item['title'].lower()
                    )
                ]
                if all(logic):
                    csv_value[7] = item['producers']
                    if item['producers'] is None:
                        continue
                    
                    print(f"Match found: {csv_value[1]} - {csv_value[3]}")
                    continue
```

```python
df = pandas.DataFrame(grammies, columns=grammies.pop(0))
```

```python
df[df.year == f'{YEAR}'][['id', 'song_or_album', 'producers']]
```

```python
df.to_csv(pathlib.Path('grammy_winners_v2.csv'), index=False)
```
