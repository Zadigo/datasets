# Apply Producers Fix

Apply the fixes made from the `fix_missed_producers.csv` to the `grammy_winners.csv` file

```python
import pandas
import pathlib
```

```python
TMP_DIR = pathlib.Path('.') / 'tmp'
```

```python
df1 = pandas.read_csv('grammy_winners.csv')
```

```python
df2 = pandas.read_csv(TMP_DIR / 'fix_missed_producers.csv')
```

```python
year = 1958
```

```python
section_to_change = df2[df2.year == year][['id', 'producers']]
```

```python
section_to_change.head()
```

```python
for item in section_to_change.itertuples(name='Award'):
    df1.loc[item.Index, 'producers'] = item.producers
```
