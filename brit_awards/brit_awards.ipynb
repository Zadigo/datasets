```python
import pandas
import pathlib
```

```python
files = list(pathlib.Path('.').glob('json/*.json'))
```

```python
dfs = [pandas.read_json(file) for file in files]
```

```python
df = pandas.concat(dfs, ignore_index=True)
```

```python
df.head()
```

## Additional columns

```python
def check_is_person(constraints: list[str]) -> callable:
    def wraped(name: str) -> bool:
        if not name:
            return False
        
        # values = ['Female', 'Male', 'Newcomer', 'Group', 'Artist']

        result = False
        for value in constraints:
            if value in name:
                result = True
                break
        return result
    return wraped
```

```python
df['is_person'] = df.details.map(check_is_person(
    [
        'Female', 
        'Male', 
        'Newcomer', 
        'Group', 
        'Artist',
        'Outstanding Contribution',
        'British Breakthrough',
        'Act'
    ]
))
```

```python
df['is_album_or_single'] = df.details.map(check_is_person(['Album', 'Single']))
```

```python
artists = df[df.winner.isna() == False & (df.winner != '')]['winner'].drop_duplicates()
```

## Normalize details

```python
def title_case(detail: str):
    if not detail:
        return detail
    
    if detail.isupper():
        return detail.lower().title()
    return detail
```

```python
df['details'] = df.details.map(title_case)
```

## Output data

```python
artists.to_csv('brit_awards_artists.csv', index=False)
```
