# Apply Producers Fix

Apply the fixes made from the `fix_missed_producers.csv` to the `grammy_winners.csv` file

```python
import pandas
import pathlib
```

```python
TMP_DIR = pathlib.Path('.') / 'tmp'
```

```python
df1 = pandas.read_csv('grammy_winners_v2.csv')
```

```python
df2 = pandas.read_csv(TMP_DIR / 'fix_missed_producers.csv')
```

```python
df2.groupby('year')['year'].count()
```

```python
for item in df2.itertuples(name='Award'):
    df1.loc[(df1.year == item.year) & (df1.id == item.id), 'producers'] = item.producers
    df1.loc[(df1.year == item.year) & (df1.id == item.id), 'artist'] = item.artist
    df1.loc[(df1.year == item.year) & (df1.id == item.id), 'song_or_album'] = item.song_or_album
```

```python
df1.to_csv('grammy_winners_v2.csv', index=False)
```

```python
df3 = pandas.read_csv('tmp/fixes.csv')
```

```python
for item in df3.itertuples():
    df1.loc[df1.id == item.line, 'producers'] = item.producers
```

```python
df1.to_csv('grammy_winners_v2.csv', index=False)
```

```python
remaining_issues = df1[df1.producers == df1.song_or_album][['id', 'category', 'url', 'year', 'producers', 'song_or_album', 'artist']]
remaining_issues['id'].describe()['count']
```

```python
remaining_issues.to_csv(TMP_DIR / 'fix_missed_producers.csv', index=False)
```
